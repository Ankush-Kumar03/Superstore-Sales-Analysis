# 7. Feature Engineering 

```python
df2 = df.copy()
```

# Order Year 

```python
df2["Order Year"] = df2["Order Date"].dt.year
```

# Order Month

```python
df2["Order Month"] = df2["Order Date"].dt.month
```

# Month Name 

```python
df2["Month Name"] = df2["Order Date"].dt.month_name()
```

# Order Quarter 

```python
df2["Order Quarter"] = df2["Order Date"].dt.quarter
```

# Shipping Days 

```python
df2["Shipping Days"] = (df2["Ship Date"] - df2["Order Date"]).dt.days
```

```python
df2[["Order Year","Order Month","Month Name","Order Quarter","Shipping Days"]].head()
```

### Observation

    -Order Month was extracted from the Order Date column to analyze monthly  sales and profit trends.
    -Order Year was created to compare business performance across different years and evaluate long-term growth trends.
    -Shipping Days was calculated as the difference between Ship Date and Order Date.

## Business Question 7

### Which Sub-Categories generate high sales but low profit?

```python
sub_category_analysis = df2.groupby("Sub-Category").agg({
           "Sales": "sum",
           "Profit": "sum"
       }).sort_values(by="Sales", ascending=False)

sub_category_analysis
```

##### Sales and profit comparisions

```python
fig, ax = plt.subplots(1, 2, figsize=(14,7))
sub_category_analysis["Sales"].head(5).plot(kind="bar", ax = ax[0])
ax[0].set_title("Top 5 Sub-Category by Sales")
ax[0].set_xlabel("Sub-category")
ax[0].set_ylabel("Sales")


sub_category_analysis["Profit"].head(5).plot(kind="bar", ax = ax[1])
ax[1].set_title("Top 5 Sub-Category by Profit")
ax[1].set_xlabel("Sub-category")
ax[1].set_ylabel("Profit")

plt.tight_layout()
```

##### Visualization

```python
fig, ax = plt.subplots(figsize=(14,7))

ax.scatter(
    sub_category_analysis["Sales"],
    sub_category_analysis["Profit"]
)

for i in sub_category_analysis.index:
    ax.text(
        sub_category_analysis.loc[i, "Sales"],
        sub_category_analysis.loc[i, "Profit"]+1000,
        i
    )

ax.set_title("Sales vs Profit by Sub-Category")
ax.set_xlabel("Sales")
ax.set_ylabel("Profit")

plt.tight_layout()
plt.show()
```

#### Observation

- **Phones** generate both high sales and high profit.
- **Tables** generate high sales but low (or negative) profit.

#### Business Insight

- Some products are generating strong revenue but are not profitable.
- This may be due to high discounts, high production costs, or pricing issues.

#### Recommendation

- Review the pricing strategy for Tables.
- Reduce unnecessary discounts.
- Analyze production or procurement costs.
- Focus on increasing the profitability of high-selling products.

## Business Question 8

### Does Discount Affect Profit?

```python
fig, ax = plt.subplots(figsize=(8,6))

ax.scatter(
    df2["Discount"],
    df2["Profit"],
    alpha = 0.6
)

ax.axhline(
    y = 0,
    color = "red",
    linewidth = 1,
    linestyle = "--"
)

ax.set_title("Discount vs Profit")
ax.set_xlabel("Discount")
ax.set_ylabel("Profit")

plt.tight_layout()
plt.show()    



correlation = df2["Discount"].corr(df2["Profit"])

print("Correlation :", correlation)
```

#### Observation

- The scatter plot shows the relationship between Discount and Profit.
- The correlation coefficient is negative.
- As discount increases, profit generally tends to decrease.
- Several high-discount orders generate low or even negative profit.

#### Business Insight

- Excessive discounts reduce overall profitability.
- Increasing discounts may improve sales volume but does not always increase profit.
- Some products become loss-making because of very high discounts.

#### Recommendation

- Avoid unnecessary high discounts.
- Apply discounts only to products with healthy profit margins.
- Regularly monitor the impact of discounts on profitability.

#### Business Insight

- Copiers appear to be a high-margin product category, contributing substantially more profit per order than other sub-categories.

#### Recommendation

## Overall Business Findings

- The company's profitability is not evenly distributed across all products and regions. While some categories and sub-categories generate strong profits, others consistently underperform.

- Technology emerged as the strongest performing product category, delivering the highest average profit per order. Within this category, Copiers generated exceptionally high average profits, followed by Accessories and Phones.

- Furniture showed relatively low profitability despite contributing to sales, indicating that higher revenue does not always translate into higher profit.

- Profitability varies significantly across geographical regions and states. Certain states consistently reported losses, suggesting location-specific challenges that require targeted business strategies.

- Customer segments receive similar average discounts, indicating a relatively balanced discount policy. However, the effectiveness of these discounts should be evaluated against sales growth and profitability.

- Shipping methods also contribute differently to profitability, suggesting that logistics decisions can influence business performance.

- Monthly sales and profit trends indicate seasonal demand, with specific months generating noticeably higher revenue and profit than others. These periods represent opportunities for focused marketing and inventory planning.

- Overall, the analysis shows that profitability is influenced by multiple factors, including product category, product mix, geography, discounts, and seasonal demand. Business decisions should therefore be based on a combination of these factors rather than a single performance metric.

## Overall Business Insights

- High sales alone are not a reliable indicator of business success. Profitability should be considered alongside sales while evaluating product performance.

- Technology products represent the company's strongest profit drivers and should remain a strategic focus for revenue growth.

- Persistent losses in specific states suggest that a uniform business strategy may not be effective across all locations. Regional pricing, logistics, and marketing strategies should be evaluated separately.

- The significant profitability gap between product categories indicates that resource allocation, inventory planning, and promotional campaigns should prioritize high-margin products.

- Seasonal sales patterns provide opportunities for better inventory management, workforce planning, and marketing campaigns during peak demand periods.

- Discount strategies should be continuously monitored to ensure that increased discounts are translating into sustainable profitability rather than simply increasing sales volume.

## Overall Business Recommendations

- Increase investment in high-profit product categories and sub-categories such as Technology, Copiers, Accessories, and Phones to maximize overall profitability.

- Conduct a detailed root-cause analysis for loss-making states by reviewing pricing strategies, operational expenses, shipping costs, and customer demand.

- Review the profitability of Furniture products and identify opportunities to improve margins through pricing optimization, supplier negotiations, or product portfolio adjustments.

- Develop region-specific business strategies instead of applying a uniform pricing and discount policy across all markets.

- Align marketing campaigns and inventory planning with seasonal demand to maximize sales during high-performing months while avoiding excess inventory during slower periods.

- Continuously monitor discount effectiveness by comparing discount levels with both sales and profit to ensure long-term business sustainability.

- Establish a KPI dashboard to track key business metrics such as Sales, Profit, Profit Margin, Discount Rate, Regional Performance, and Product Performance on a regular basis.
